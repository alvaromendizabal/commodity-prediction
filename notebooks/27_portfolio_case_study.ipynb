## Problem and contribution
Rank multi-horizon returns across 424 commodity-related targets while respecting what was knowable at prediction time. The engineering challenge combines delayed outcomes, heterogeneous market histories, noisy ranking signals, and temporal instability.

The project spans data contracts, availability-aware representations, chronological validation, pooled nonlinear modeling, controlled candidate experiments, checkpointed AWS execution, and analytical communication. It is a completed research case study, not a deployed trading system.

The public evidence emphasizes judgment: candidate definitions were frozen, evaluations used matching target/date populations, and negative or inconclusive results were retained. Complete reproduction of all leading competition systems is not claimed.

## Diagnose the experiment before adding compute
The original mixed-horizon panel scored 0.418181 against 0.403381 on its screening period. A 106-target one-day expansion scored 0.385559. A zero-fit reconciliation established that all 37 shared targets had identical predictions. The change came from removing 27 longer-horizon replacements and adding 69 one-day replacements.

This is order-dependent accounting, not causal feature attribution. It justified testing the original candidate on later periods, rather than selecting a favorable subset after observing outcomes. The separate 535-date development reference is **0.309709**; it must not be ranked directly against scores measured on different dates.

## A positive estimate is not automatic promotion
On 355 later development dates and all 424 scored targets, the frozen panel scored **0.289365** against **0.276025**. Both individual period differences were positive. The pooled difference was **+0.013340**, with a conditional paired 95% interval of **[-0.009061, +0.035750]**.

The predeclared gate required a positive lower interval endpoint. It was not met, so the incumbent was retained. These were previously used development periods: retrospective temporal replication, not an untouched final test.

## Complete the assessment and report the control
Before final access, the incumbent reconstruction exactly replayed archived development predictions for 175 dates and 424 targets. The final model and assessment contract were then locked.

On the documented historical reservation of **247 dates, IDs 1714-1960**, the frozen incumbent scored **0.190453** versus **0.212085** for the training-only historical mean. All 424 targets were included and no requested dates were excluded from the primary score.

The difference was **-0.021632**, with a conditional paired 95% interval of **[-0.133651, +0.080336]**. There was no demonstrated advantage. The outcome was recorded without final-period model refitting or post-assessment reselection. Access outside the available evidence cannot be independently excluded; this is a documented historical assessment, not a claim of Kaggle hidden-test performance.

## Engineering judgment and limits
The work produced a functioning point-in-time research workflow, controlled model comparisons, a zero-fit explanation for an apparent regression, and a completed final assessment. It also showed that additional complexity did not demonstrate a final advantage over the control.

This portfolio does not claim a competition win, a complete reproduction of leading solutions, a live deployment, or trading profitability. The research cycle is closed; new modeling would need a separately defined evaluation design.
