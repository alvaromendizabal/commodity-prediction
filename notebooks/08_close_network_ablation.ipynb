## 2. Mechanism, timing and overlap

Two comparable graphs use 126 prior rows and five incoming peers, refreshed every 21 rows. Synchronous weights compare returns at the same prior row. Directed weights compare follower return at s with peer return at s−1. At refresh t, follower history ends at t−1. **Both prediction legs are excluded from their own peer pool.** Correlations are signed; missing actual observations are not silently set to zero.

The 12 numerical columns describe current peer impulse, a five-row peer mean, and previous peer impulse relative to current own response, each projected as a target-leg difference and sum. Four availability columns control for network/observation coverage.

Existing global ranks, PCA, currency-identity graphs, proxy betas and close-pair risk features are not new discoveries. The distinct hypothesis is leave-pair-out **incoming raw-price network information**. It is not a verified causal network or a target-label prior.

[Bennett, Cucuringu & Reinert](https://arxiv.org/abs/2201.08283) motivates this line of research in equities. This is not a replication or proof of commodity-return predictability. Read `docs/manual_close_network.md` for the exact experimental and source-availability contract.
